## Lesson Plan

1. Cumulative Distribution Functions (CDFs): $F_X(s) = P(X \leq x)$
2. Probability Density Functions (PDFS): $f_X(x) = \frac{d}{dx}F_X(x)$
3. Expectation values and moments: $E[X] = \langle X \rangle = \int_{-\infty}^{+\infty}xf_X(x)dx$
4. Common distributions (uniform, normal, etc.)

Throughout, we will also do a few practice problems (as time permits, but ideally exercises 1, 4, and 5 from the attached handout.)
